```python
# Import required libraries
import pandas as pd
import numpy as np

# Upload the dataset file
from google.colab import files
uploaded = files.upload()

# Read the uploaded CSV file
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

# Display first 5 rows
print(df.head())

# Display number of rows and columns
print("Shape of dataset:", df.shape)
```

```python
# Check missing values in each column
print("Missing values:")
print(df.isnull().sum())
```

```python
# Remove loan_id because it is only an identification number
# Separate input features (X) and target (Y)

X = df.drop(['loan_id', ' loan_status'], axis=1)
y = df[' loan_status']

# Display features
print("Features:")
print(X.head())

# Display target
print("\nTarget:")
print(y.head())
```

```python
# Import LabelEncoder
from sklearn.preprocessing import LabelEncoder

# Convert categorical feature columns into numerical values
for col in X.columns:
    if X[col].dtype == 'object':
        le = LabelEncoder()
        X[col] = le.fit_transform(X[col])

# Remove extra spaces from target values
y = y.str.strip()

# Convert Approved to 1 and Rejected to 0
y = y.map({
    'Approved': 1,
    'Rejected': 0})

# Display encoded features
print("Encoded Features:")
print(X.head())

# Display encoded target
print("\nEncoded Target:")
print(y.head())
```

```python
# Find total number of records
n = len(X)

# Calculate K using square root of N
k = int(np.sqrt(n))

print("Number of records (N):", n)
print("K using √N:", k)
```

```python
# Function to calculate Manhattan distance
def manhattan_distance(row1, row2):

    # Calculate absolute difference between each feature and add all the differences
    return np.sum(np.abs(row1 - row2))
```

```python
# Function to calculate distances from test data to all other records

def knn_predict(X, y, test_index, k):

    distances = []

    # Get the test record
    test_data = X.iloc[test_index]

    # Calculate distance with every record
    for i in range(len(X)):

        # Do not compare the record with itself
        if i == test_index:
            continue

        # Calculate Manhattan distance
        distance = manhattan_distance(
            X.iloc[i].values,
            test_data.values)

        # Store distance and target value
        distances.append((distance, y.iloc[i]))

    return distances
```

```python
# Select the first record as test data
test_index = 0

# Calculate distances
distances = knn_predict(
    X, y, test_index, k)

# Sort distances from smallest to largest
distances.sort(key=lambda x: x[0])

# Display first 10 smallest distances
print("Smallest distances:")
print(distances[:10])
```

```python
# Select K records with the smallest distances
nearest_neighbors = distances[:k]

# Display K nearest neighbors
print("K nearest neighbors:")
print(nearest_neighbors)
```

```python
# Extract target values of K nearest neighbors
neighbor_values = [value for distance, value in nearest_neighbors]

# Calculate average of target values
average = np.mean(neighbor_values)

# Display average
print("Average:", average)
```

```python
# Predict loan status using the average

if average >= 0.5:
    prediction = 1
    result = "Approved"
else:
    prediction = 0
    result = "Rejected"

# Display prediction
print("Predicted Output:", prediction)
print("Predicted Loan Status:", result)
```

```python
# Get actual output of the selected record
actual = y.iloc[test_index]

# Convert numerical value to text
if actual == 1:
    actual_result = "Approved"
else:
    actual_result = "Rejected"

# Display actual output
print("Actual Output:", actual)
print("Actual Loan Status:", actual_result)
```

```python
# Create empty list for storing predictions
predictions = []

# Predict first 500 records
for i in range(500):

    # Calculate distances from current record
    distances = knn_predict(X, y, i, k)

    # Sort distances
    distances.sort(key=lambda x: x[0])

    # Select K nearest neighbors
    nearest_neighbors = distances[:k]

    # Get target values
    neighbor_values = [value for distance, value in nearest_neighbors]

    # Calculate average of neighbor target values
    average = np.mean(neighbor_values)

    # Predict output
    if average >= 0.5:
        prediction = 1
    else:
        prediction = 0

    # Store prediction
    predictions.append(prediction)

print("Predictions generated successfully.")
print("Number of predictions:", len(predictions))
```

```python
# Create a table of actual and predicted values
result_df = pd.DataFrame({
    'Actual': y.iloc[:500].values,
    'Predicted': predictions})

# Display first 50 results
print(result_df.head(50))
```

```python
# Count correct predictions
correct = np.sum(y.iloc[:500].values == np.array(predictions))

# Calculate accuracy
accuracy = (correct / len(predictions)) * 100

# Display results
print("Correct Predictions:", correct)
print("Total Predictions:", len(predictions))
print("Accuracy:", accuracy, "%")
```
